```python
# import modules
import gmag.arrays.carisma as carisma
import pandas as pd

from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"
```

```python
sd = '2015-01-01'
ed = '2015-01-31'

d_ser = pd.date_range(start=pd.to_datetime(sd), 
                              end=pd.to_datetime(ed),
                              freq='D')


d_ser[0]
d_ser[-1]
```

```python
d_ser[0].strftime("%Y%m%d")
```

```python
fn = r'D:\scratch\gill_2_'

ctype = 'zlib'

for dt in d_ser: 
    ds = dt.strftime("%Y%m%d")
    df, meta = carisma.load(sdate=dt,drop_flag=False)

    for col in df.select_dtypes(include='float'):
        df[col] = df[col].astype('float32')

    df.to_hdf(f'{fn}_{ctype}.hdf',key=f'gill_{ds}',complib=ctype, complevel=3, format='table')
```
